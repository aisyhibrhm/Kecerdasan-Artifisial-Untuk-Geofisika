# Modul: Menyusun Prompt AI yang Baik untuk Permasalahan Geofisika

## A. Pendahuluan

Dalam kegiatan geofisika seperti pengolahan data, pemodelan, dan interpretasi, prompt untuk AI tidak hanya berfungsi sebagai pertanyaan, tetapi juga sebagai petunjuk kerja yang terstruktur. Prompt yang disusun dengan baik membantu AI memberikan analisis yang lebih terarah serta memudahkan pengguna memahami dasar dari setiap proses yang dilakukan.

Struktur prompt dapat disusun menggunakan beberapa komponen utama yang membentuk alur kerja, yaitu R-T-C-I-K-O-Q-V (Role, Task, Context, Input, Constraints, Output, Quality Criteria, dan Optional Verification).

## B. Penjelasan Tiap Komponen

Framework **R-T-C-I-K-O-Q-V** terdiri atas beberapa bagian yang saling melengkapi untuk membentuk prompt yang terarah. Setiap komponen memiliki fungsi yang berbeda dalam membantu AI memahami tugas dan menghasilkan respons sesuai kebutuhan.

**2.1 Role (R)**

Menentukan **peran atau keahlian AI** dalam menjalankan tugas. Role membantu AI menyesuaikan sudut pandang dan pengetahuan yang digunakan sesuai dengan permasalahan yang diberikan.

**2.2 Task (T)**

Menjelaskan **tugas utama yang harus dilakukan oleh AI**. Bagian ini harus dibuat secara jelas agar AI memahami tujuan analisis yang ingin dicapai.

**2.3 Context (C)**

Memberikan **latar belakang atau kondisi permasalahan** yang sedang dianalisis. Context membantu AI memahami situasi penelitian sehingga analisis tidak dilakukan tanpa mempertimbangkan kondisi yang ada.

**2.4 Input (I)**

Berisi **data atau informasi yang diberikan kepada AI** sebagai dasar untuk melakukan analisis. Input dapat berupa data pengukuran, parameter, tabel, grafik, maupun informasi pendukung lainnya.

**2.5 Constraints (K)**

Menentukan **batasan dalam proses analisis**. Bagian ini digunakan untuk mencegah AI membuat asumsi atau memberikan informasi yang tidak didukung oleh data yang tersedia.

**2.6 Output (O)**

Menjelaskan **bentuk keluaran yang diharapkan dari AI**. Output dapat berupa narasi, tabel, interpretasi, langkah analisis, atau bentuk lainnya sesuai tujuan pekerjaan.

**2.7 Quality Criteria (Q)**

Berisi **kriteria yang digunakan untuk memastikan kualitas jawaban AI**. Kriteria ini dapat mencakup ketepatan konsep, kesesuaian dengan data, kejelasan penjelasan, dan konsistensi hasil analisis.

**2.8 Optional Verification (V)**

Merupakan bagian **pemeriksaan tambahan** yang dilakukan apabila diperlukan. Verifikasi dapat digunakan untuk membandingkan hasil analisis dengan data pendukung agar diperoleh hasil yang lebih konsisten.

## 1. STUDI KASUS SEISMIK REFRAKSI UNTUK IDENTIFIKASI BATUAN DASAR
**Role**

Anda berperan sebagai seorang Geophysicist Exploration yang memiliki kompetensi dalam metode seismik refraksi, analisis waktu tiba gelombang, penentuan kecepatan lapisan, serta interpretasi struktur bawah permukaan. Anda mampu menghubungkan variasi kecepatan gelombang seismik dengan karakteristik material bawah permukaan.

**Task**

Lakukan analisis terhadap data seismik refraksi untuk memperkirakan susunan lapisan bawah permukaan dan kedalaman batuan dasar. Analisis mencakup hubungan antara waktu tiba gelombang dengan jarak sumber-geophone, penentuan kecepatan setiap lapisan, serta estimasi kedalaman batas antar lapisan berdasarkan data yang tersedia.

**Context**

Survei seismik refraksi dilakukan pada suatu lintasan untuk mengetahui kondisi bawah permukaan sebelum pembangunan infrastruktur. Data diperoleh menggunakan beberapa geophone yang ditempatkan dengan jarak tertentu dan sumber getaran pada beberapa posisi. Hasil pengukuran berupa waktu tiba gelombang P yang selanjutnya dapat dianalisis menggunakan kurva travel time untuk mengetahui variasi kecepatan terhadap kedalaman.

**Input**

Data yang diberikan berupa posisi geophone, jarak antar geophone, posisi sumber, waktu tiba gelombang pertama, panjang lintasan, serta informasi geologi daerah penelitian apabila tersedia. Apabila terdapat grafik travel time atau hasil picking gelombang, gunakan informasi tersebut sebagai dasar tambahan dalam interpretasi.

**Constraints**

Interpretasi tidak boleh menentukan kedalaman atau kecepatan lapisan tanpa berdasarkan data. Jangan menganggap jumlah lapisan tertentu apabila belum didukung oleh pola travel time. Bedakan hasil pengukuran, hasil perhitungan, asumsi, dan interpretasi. Jenis batuan juga tidak boleh ditentukan secara pasti hanya berdasarkan nilai kecepatan tanpa informasi geologi pendukung.

**Output**

Sajikan hasil analisis secara sistematis dengan menjelaskan karakteristik data seismik, pola hubungan waktu tiba dan jarak, perkiraan kecepatan setiap lapisan, kemungkinan batas antar lapisan, perkiraan kedalaman batuan dasar apabila memungkinkan, serta tingkat ketidakpastian interpretasi. Gunakan bahasa akademik dan terminologi seismik yang sesuai.

## 2. STUDI KASUS METODE MAGNETIK UNTUK PEMETAAN STRUKTUR GEOLOGI
**Role**

Anda berperan sebagai seorang Geophysicist yang berpengalaman dalam metode geomagnetik, pengolahan anomali medan magnet, transformasi data, serta interpretasi struktur geologi bawah permukaan. Anda mampu menghubungkan pola anomali magnetik dengan kemungkinan keberadaan kontak litologi, sesar, atau struktur geologi lainnya.

**Task**

Analisis data magnetik untuk mengidentifikasi pola anomali dan memperkirakan keberadaan struktur geologi di daerah penelitian. Analisis meliputi pengamatan distribusi anomali, perubahan nilai magnetik, pola gradien, serta hubungan antara anomali dengan informasi geologi yang tersedia.

**Context**

Survei geomagnetik dilakukan pada suatu wilayah yang memiliki indikasi keberadaan struktur geologi yang belum diketahui secara jelas. Data medan magnet hasil pengukuran kemudian melalui beberapa tahapan pengolahan untuk memperoleh anomali magnetik yang lebih representatif. Hasil pengolahan tersebut digunakan untuk membantu menggambarkan pola struktur bawah permukaan.

**Input**

Data yang tersedia dapat berupa koordinat titik pengukuran, nilai medan magnet total, data base station, hasil koreksi variasi harian, data IGRF, peta anomali magnetik, serta informasi geologi permukaan. Hasil transformasi seperti RTP, upward continuation, atau turunan horizontal dapat digunakan apabila tersedia.

**Constraints**

Jangan menentukan keberadaan sesar atau kontak litologi hanya berdasarkan satu pola anomali. Hindari menganggap anomali tinggi selalu menunjukkan batuan tertentu dan anomali rendah selalu menunjukkan jenis batuan tertentu. Setiap interpretasi harus mempertimbangkan posisi, bentuk, amplitudo, dan pola anomali serta data geologi pendukung.

**Output**

Sajikan analisis berupa uraian mengenai karakteristik anomali magnetik, zona perubahan nilai yang signifikan, kemungkinan keberadaan struktur geologi, hubungan anomali dengan kondisi geologi, serta tingkat keyakinan interpretasi. Jika data belum cukup untuk menentukan jenis batuan atau struktur secara pasti, jelaskan keterbatasannya.

## 3. STUDI KASUS GRAVITASI UNTUK IDENTIFIKASI STRUKTUR BAWAH PERMUKAAN
**Role**

Anda berperan sebagai seorang Geophysicist Exploration yang memiliki keahlian dalam metode gravitasi, pengolahan data anomali Bouguer, analisis densitas, serta interpretasi struktur bawah permukaan berdasarkan variasi medan gravitasi.

**Task**

Lakukan analisis data gravitasi untuk mengetahui pola variasi densitas bawah permukaan dan mengidentifikasi kemungkinan struktur geologi atau batas litologi. Analisis harus mempertimbangkan distribusi anomali gravitasi, pola regional dan residual, serta informasi geologi yang tersedia.

**Context**

Survei gravitasi dilakukan pada suatu daerah yang memiliki variasi kondisi geologi bawah permukaan. Data gravitasi hasil pengukuran telah melalui koreksi yang diperlukan untuk memperoleh nilai anomali yang dapat digunakan dalam interpretasi. Pemisahan anomali regional dan residual dapat dilakukan untuk membantu mengidentifikasi sumber anomali dengan skala yang berbeda.

**Input**

Data yang diberikan dapat berupa koordinat titik pengukuran, elevasi, nilai gravitasi, nilai anomali Bouguer, densitas batuan, peta anomali gravitasi, serta informasi geologi daerah penelitian. Apabila tersedia, gunakan hasil regional-residual atau turunan data gravitasi sebagai informasi tambahan.

**Constraints**

Jangan menghubungkan anomali gravitasi secara langsung dengan satu jenis batuan tanpa mempertimbangkan densitas dan kondisi geologi. Hindari membuat model bawah permukaan dengan kedalaman atau densitas yang tidak didukung oleh data. Interpretasi harus mempertimbangkan sifat ambiguitas metode gravitasi dan kemungkinan adanya lebih dari satu sumber anomali.

**Output**

Jelaskan karakteristik pola anomali gravitasi, perbedaan antara zona anomali tinggi dan rendah, kemungkinan variasi densitas bawah permukaan, serta struktur geologi yang dapat ditafsirkan dari pola tersebut. Sertakan penjelasan mengenai keterbatasan interpretasi dan data tambahan yang diperlukan apabila model bawah permukaan belum dapat ditentukan secara jelas.

## 4. STUDI KASUS GPR UNTUK IDENTIFIKASI STRUKTUR BAWAH PERMUKAAN DANGKAL
**Role**

Anda berperan sebagai seorang Geophysicist Near Surface yang memiliki kemampuan dalam metode Ground Penetrating Radar (GPR), analisis radargram, pengolahan sinyal, serta interpretasi objek dan struktur bawah permukaan dangkal.

**Task**

Analisis data GPR untuk mengidentifikasi pola reflektor dan kemungkinan objek atau struktur bawah permukaan dangkal. Analisis mencakup karakteristik refleksi, kontinuitas reflektor, kedalaman objek, serta hubungan pola radargram dengan kondisi geologi atau target penelitian.

**Context**

Survei GPR dilakukan pada daerah yang membutuhkan informasi mengenai kondisi bawah permukaan dangkal. Pengukuran dilakukan menggunakan antena dengan frekuensi tertentu sepanjang lintasan yang telah ditentukan. Data hasil akuisisi kemudian diproses untuk meningkatkan kualitas radargram dan memperjelas reflektor bawah permukaan.

**Input**

Data yang tersedia berupa radargram, posisi lintasan, frekuensi antena, waktu tempuh gelombang, parameter akuisisi, serta informasi kondisi permukaan dan geologi daerah penelitian. Hasil pengolahan seperti background removal, gain, filtering, dan time-to-depth conversion dapat digunakan apabila tersedia.

**Constraints**

Jangan menentukan jenis objek hanya berdasarkan bentuk reflektor tanpa mempertimbangkan konteks geologi dan parameter akuisisi. Jangan menganggap semua reflektor kuat sebagai objek buatan manusia. Kedalaman objek tidak boleh ditentukan tanpa mempertimbangkan kecepatan gelombang elektromagnetik pada medium.

**Output**

Sajikan analisis mengenai karakteristik radargram, reflektor yang dianggap penting, kemungkinan posisi objek atau batas lapisan, perkiraan kedalaman apabila parameter kecepatan tersedia, serta hubungan hasil interpretasi dengan kondisi bawah permukaan. Jelaskan pula bagian data yang masih memiliki ketidakpastian.

## 5. STUDI KASUS METODE MAGNETOTELLURIK (MT) UNTUK IDENTIFIKASI SISTEM PANAS BUMI
**Role**

Anda berperan sebagai seorang Geophysicist Geothermal yang memiliki kompetensi dalam metode Magnetotellurik (MT), analisis resistivitas bawah permukaan, interpretasi struktur geologi, dan identifikasi karakteristik sistem panas bumi.

**Task**

Analisis data MT untuk mengetahui distribusi resistivitas bawah permukaan dan mengidentifikasi zona yang berpotensi berkaitan dengan sistem panas bumi. Analisis mencakup perubahan resistivitas terhadap kedalaman, keberadaan zona konduktif atau resistif, serta hubungan hasil inversi dengan kondisi geologi.

**Context**

Survei MT dilakukan pada daerah yang memiliki indikasi aktivitas panas bumi. Data MT diperoleh pada beberapa titik pengukuran dengan rentang frekuensi tertentu dan kemudian diproses melalui tahap inversi untuk menghasilkan model resistivitas bawah permukaan. Model tersebut digunakan untuk membantu memahami struktur sistem panas bumi.

**Input**

Data yang diberikan dapat berupa koordinat titik MT, data apparent resistivity, phase, frekuensi, hasil inversi 1D atau 2D, penampang resistivitas, RMS error, serta informasi geologi dan manifestasi panas bumi apabila tersedia.

**Constraints**

Jangan menetapkan suatu zona sebagai reservoir panas bumi hanya berdasarkan nilai resistivitas. Zona konduktif dapat disebabkan oleh beberapa faktor seperti lempung, fluida, atau kondisi geologi tertentu. Interpretasi harus mempertimbangkan kedalaman, kontinuitas zona, informasi geologi, manifestasi permukaan, serta kualitas hasil inversi.

**Output**

Sajikan hasil analisis mengenai distribusi resistivitas terhadap kedalaman, zona konduktif dan resistif yang menonjol, kemungkinan hubungan zona tersebut dengan sistem panas bumi, serta interpretasi struktur bawah permukaan. Jelaskan tingkat keyakinan dan ketidakpastian interpretasi berdasarkan kualitas data dan hasil inversi.
